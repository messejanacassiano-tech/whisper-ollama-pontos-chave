```python
from IPython.display import Audio

Audio("audio_teste.mp3")
```

```python
import whisper

modelo = whisper.load_model("base")

resultado = modelo.transcribe("audio_teste.mp3", language="pt")
texto_transcrito = resultado["text"].strip()

print("Transcrição:")
print(texto_transcrito)
```

```python
import ollama

prompt = f"""
Analise o texto transcrito abaixo e extraia os principais pontos-chave em tópicos curtos, objetivoe e em português:

Texto:{texto_transcrito}"""

resposta = ollama.chat(
    model="llama3.2",
    messages=[{"role": "user", "content": prompt}],
)

pontos_chave = resposta["message"]["content"].strip()

print("--- Pontos-chave ---")
print(pontos_chave)
```
